```python
# Lista ejecutable de 6 títulos de canciones de tipo string.
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
```

```python
# Lista ejecutable con el conteo de reproducciones asociadas a cada canción.
playcounts = [78, 29, 44, 21, 89, 5]
```

```python
# Crea el diccionario 'plays' en tiempo de ejecución combinando las listas 'songs' y 'playcounts' con zip() y dict comprehension.
plays = {key:value for key, value in zip(songs, playcounts)}
```

```python
# Imprime el diccionario 'plays' en la consola.
print(plays)
```

```python
# Agrega "Purple Haze" con valor 1 al diccionario ejecutable 'plays' mediante .update().
plays.update({"Purple Haze": 1})
```

```python
# Actualiza la clave "Respect" asignándole 94 reproducciones mediante .update().
plays.update({"Respect": 94})
```

```python
# Imprime el diccionario 'plays' modificado precedido por "After: ".
print("After: ", plays)
```

```python
# Crea el diccionario anidado 'library' usando el diccionario 'plays' en la clave "The Best Songs" y un diccionario vacío `{}` en "Sunday Feelings".
library = {"The Best Songs": plays, "Sunday Feelings": {}}
```

```python
# Imprime la estructura final del diccionario anidado 'library' en la consola.
print(library)
```
